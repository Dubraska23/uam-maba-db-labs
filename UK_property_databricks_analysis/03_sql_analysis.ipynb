```sql
-- Ver todas las tablas disponibles
SHOW TABLES IN ucm;
```

```sql
-- ============================================
-- QUERY 1: Identificar qué ciudades tienen mayor precio medio y volumen
-- ============================================

SELECT 
    town_city,
    total_transactions,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(min_price, 0) AS min_price,
    ROUND(max_price, 0) AS max_price
FROM ucm.gold_city_prices
WHERE total_transactions > 50
ORDER BY avg_price DESC
LIMIT 10;
```

```sql
-- ============================================
-- QUERY 2: Analizar la tendencia del mercado inmobiliario en cuanto a precios
-- ============================================
SELECT 
    year,
    month,
    total_transactions,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(std_price, 0) AS price_volatility,
    ROUND(total_value, 0) AS total_market_value
FROM ucm.gold_temporal_trends
ORDER BY year, month;
```

```sql
-- ============================================
-- QUERY 3: Analizar qué tipo de vivienda se vende más
-- ============================================
SELECT 
    property_type_desc,
    property_age,
    tenure_type,
    total_sales,
    ROUND(total_price, 0) AS total_market_value,
    ROUND(avg_price, 0) AS avg_price
FROM ucm.gold_property_analysis
WHERE total_sales > 20
ORDER BY total_sales DESC;
```

```sql
-- ============================================
-- QUERY 4: Identificar diferencias de precios dentro de distritos
-- ============================================
SELECT 
    district,
    total_transactions,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(min_price, 0) AS min_price,
    ROUND(max_price, 0) AS max_price
FROM ucm.gold_district_analysis
WHERE total_transactions > 50
ORDER BY avg_price DESC
LIMIT 15;
```

```sql
-- ============================================
-- QUERY 5: Analizar y entender la distribución del mercado inmobiliario por categorías de precios
-- ============================================
SELECT 
    price_category,
    total_properties,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(min_price, 0) AS min_price,
    ROUND(max_price, 0) AS max_price,
    ROUND((total_properties * 100.0 / SUM(total_properties) OVER()), 2) AS percentage_market
FROM ucm.gold_price_category_analysis
ORDER BY avg_price DESC;
```
