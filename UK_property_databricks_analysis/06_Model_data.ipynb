# Model Data

Objetivo: crear una tabla con los datos necesarios para entrenar un modelo que prediga la evolución de los precios en Londres.

```sql
%python
df_model_data = spark.sql("""
SELECT
  CAST(CONCAT(year, '-', LPAD(month, 2, '0'), '-01') AS DATE) AS date_of_transfer,
  ROUND(AVG(price), 1) AS avg_price
FROM
  workspace.uk_housing.silver_property_sales
WHERE town_city = 'LONDON'
GROUP BY year, month;
""")

# Guardar
df_model_data.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.uk_housing.model_data")

print("\n Tabla datos del modelo creada: model_data")
print("\nÚltimos 12 meses:")
display(df_model_data.orderBy("date_of_transfer", ascending=False).limit(12))
```
