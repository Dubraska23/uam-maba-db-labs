```sql
-- ============================================
-- QUERY 1: Top 20 ciudades más caras
-- ============================================

SELECT 
    town_city,
    total_transactions,
    ROUND(avg_price, 0) as avg_price_gbp,
    ROUND(median_price, 0) as median_price_gbp,
    ROUND(min_price, 0) as min_price_gbp,
    ROUND(max_price, 0) as max_price_gbp
FROM gold_city_prices
WHERE town_city != 'Unknown'
ORDER BY avg_price DESC
LIMIT 20;
```
