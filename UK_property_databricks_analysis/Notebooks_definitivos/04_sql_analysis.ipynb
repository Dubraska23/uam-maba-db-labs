```sql
-- ============================================
-- QUERY 1: Identificar qué ciudades tienen mayor volumen de transacciones y comparar precios
-- ============================================

SELECT 
    town_city,
    total_transactions,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(min_price, 0) AS min_price,
    ROUND(max_price, 0) AS max_price
FROM workspace.default.gold_city_prices
ORDER BY total_transactions DESC
LIMIT 10;
```

La ciudad con mayor volumen de transacciones es Londres, seguida por Manchester y Bristol

```sql
-- ============================================
-- QUERY 2: Analizar la tendencia del mercado inmobiliario en cuanto a precios
-- ============================================
SELECT 
    year,
    month,
    SUM(total_transactions) AS total_transactions,
    ROUND(AVG(avg_price), 0) AS avg_price,
    ROUND(AVG(std_price), 0) AS price_volatility,
    ROUND(SUM(total_value), 0) AS total_market_value
FROM workspace.default.gold_temporal_trends
GROUP BY year, month
ORDER BY year, month;
```

El precio medio de la vivienda en Reino Unido desde 2023 a 2025 ha bajado, aunque con muchas fluctuaciones.

### Análisis de LONDON (la ciudad con más transacciones)

```sql
-- ============================================
-- QUERY 3: Analizar qué tipo de vivienda se vende más y sus precios
-- ============================================
SELECT 
    property_type_desc,
    property_age,
    tenure_type,
    total_transactions,
    ROUND(total_price, 0) AS total_market_value,
    ROUND(avg_price, 0) AS avg_price
FROM workspace.default.gold_property_analysis
WHERE town_city = 'LONDON'
ORDER BY total_transactions DESC;
```

```sql
-- ============================================
-- QUERY 4: Identificar diferencias de precios dentro de los distritos de Londres
-- ============================================
SELECT 
    district,
    total_transactions,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(min_price, 0) AS min_price,
    ROUND(max_price, 0) AS max_price
FROM workspace.default.gold_district_analysis
WHERE total_transactions > 50 AND town_city = 'LONDON'
ORDER BY avg_price DESC
LIMIT 15;
```

```sql
-- ============================================
-- QUERY 5: Analizar y entender la distribución del mercado inmobiliario por categorías de precios en Londres
-- ============================================
SELECT 
    price_category,
    total_transactions,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(min_price, 0) AS min_price,
    ROUND(max_price, 0) AS max_price,
    ROUND((total_transactions * 100.0 / SUM(total_transactions) OVER()), 2) AS percentage_market
FROM workspace.default.gold_price_category_analysis
WHERE town_city = 'LONDON'
ORDER BY avg_price DESC;
```
